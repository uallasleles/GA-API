
````python
# Estrutura de resposta dada
response_data = {
    "Response": {
        "StatusCode": 201,
        "Message": "OperationKey: a0c97908-c550-42f3-94b9-1362699ffa50"
    }
}

# 1. Acessar o valor da chave 'Message' dentro de 'Response'
message_text = response_data["Response"]["Message"]

# 2. (Opcional) Se precisar apenas do UUID, você pode limpar a string
operation_key = message_text.replace("OperationKey: ", "")

print(f"Mensagem Completa: {message_text}")
print(f"Apenas a Chave: {operation_key}")
````

````python
import json

json_string = '{"Response": {"StatusCode": 201, "Message": "OperationKey: a0c97908-c550-42f3-94b9-1362699ffa50"}}'
data = json.loads(json_string) # Converte string para dicionário
key = data["Response"]["Message"].split(": ")[1]
print(key)
````
